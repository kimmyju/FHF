# FHF: offline Kaggle submission
Attach FHF source and trained checkpoint as Kaggle datasets. **Disable Internet.** Edit SOURCE_DIR and CONFIG_FILE for your Kaggle mounts.
The YAML contains every input path, output path and batch size. A family-only checkpoint is intentionally rejected.


```python
import os, sys
# Update only these two dataset-mount locations as needed.
SOURCE_DIR = '/kaggle/input/fhf-source'
CONFIG_FILE = os.path.join(SOURCE_DIR, 'configs/kaggle.yaml')
sys.path.insert(0, os.path.join(SOURCE_DIR, 'src'))
sys.path.insert(0, os.path.join(SOURCE_DIR, 'scripts'))
from fhf.config import load_config, resolve_path
from predict import submit
cfg = load_config(CONFIG_FILE)
submit(resolve_path(cfg, 'checkpoint_dir'), resolve_path(cfg, 'test_csv'),
       resolve_path(cfg, 'submission_parquet'),
       batch_size=cfg['inference']['batch_size'], max_length=cfg['model']['max_length'])
```
